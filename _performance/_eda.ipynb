### Assumptions Testing

#### Kolmogorov-Smirnov test for normality

```python
# Test for normality of differences
metrics = [
    "Faithfulness Cosine",
    "Faithfulness ROUGE-L",
    "Answer Relevancy",
    "Context Precision",
    "Response Time"
]

for metric in metrics:
    differences = df[f'{metric} (with RAG)'] - df[f'{metric} (without RAG)']
    stat, p_value = stats.shapiro(differences)
    print(f'{metric}: Shapiro-Wilk Test Statistic={stat}, p-value={p_value}')
```

#### Homogeneity of Variance (Levene's Test)

```python
# Test for equal variance (homogeneity of variance)
for metric in metrics:
    stat, p_value = stats.levene(df[f'{metric} (with RAG)'], df[f'{metric} (without RAG)'])
    print(f'{metric}: Levene Test Statistic={stat}, p-value={p_value}')
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# Assuming 'df' is the dataframe containing your data

# Step 3: Assumptions Testing

metrics = [
    "Faithfulness Cosine",
    "Faithfulness ROUGE-L",
    "Answer Relevancy",
    "Context Precision",
    "Response Time"
]

# 3.1 Shapiro-Wilk test for normality
print("Normality Test (Shapiro-Wilk):")
for metric in metrics:
    differences = df[f'{metric} (with RAG)'] - df[f'{metric} (without RAG)']
    stat, p_value = stats.shapiro(differences)
    print(f'{metric}: Shapiro-Wilk Test Statistic={stat}, p-value={p_value}')
print()

# 3.2 Levene's test for homogeneity of variance
print("Homogeneity of Variance Test (Levene's):")
for metric in metrics:
    stat, p_value = stats.levene(df[f'{metric} (with RAG)'], df[f'{metric} (without RAG)'])
    print(f'{metric}: Levene Test Statistic={stat}, p-value={p_value}')
print()

# Step 4: Paired t-test
print("Paired t-test Results:")
for metric in metrics:
    stat, p_value = stats.ttest_rel(df[f'{metric} (with RAG)'], df[f'{metric} (without RAG)'])
    print(f'{metric}: Paired t-test Statistic={stat}, p-value={p_value}')
print()

# Step 5: Calculate Means
print("Mean and Mean Difference Calculation:")
for metric in metrics:
    mean_with_rag = df[f'{metric} (with RAG)'].mean()
    mean_without_rag = df[f'{metric} (without RAG)'].mean()
    mean_difference = mean_with_rag - mean_without_rag
    print(f'{metric}: Mean with RAG = {mean_with_rag}, Mean without RAG = {mean_without_rag}, Mean Difference = {mean_difference}')
```
